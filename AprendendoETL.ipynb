```python
import pandas as pd

datas_music = '/content/playlist_hits.json'
df_playlist = pd.read_json(datas_music)

display(df_playlist.head())
```

```python
#Construir uma nova coluna no df[Categoria_tamanho]
#sempre que a musica for maior que 30s é grande se não é pequena

# 1* Colocar todos os valores da coluna
list_tamanho_music = []

# 2* Transformar um coluna em lista
list_duracoes = df_playlist['duracao_seg'].tolist()

# 3* if duracao < 300 salvar na list_tamanho_musica = 'musica grande' if < 300 = 'musica curta'

for i in list_duracoes:
  if i < 300:
    list_tamanho_music.append('musica grande')
  else:
    list_tamanho_music.append('musica curta')

# 4* Create new column, so colocar o nome de ua coluna que não existe que ele cria
df_playlist['categoria_tamanho'] = list_tamanho_music

display(df_playlist)
```

```python
#  Tirar métricas de um datadf com .nome_metrica()

#valor medio de todos streams
media_stream = df_playlist['streams_milhoes']
display(media_stream)
display(media_stream.mean())
display(media_stream.sum())
```

```python
# Criar um filtro simples
df_dados_musica_pop = df_playlist[df_playlist['genero'] == 'Pop']
display(df_dados_musica_pop)

#Total de streams
print(f'O total de streams é de: {df_playlist['streams_milhoes'].sum()}')


# Total de streams do genero rock
df_dados_musica_pop = df_playlist[df_playlist['genero'] == 'Rock']
print(f'O total de streams é de: {df_dados_musica_pop['streams_milhoes'].sum()}')
```

```python
#Filtro simples, mas com 2 criterios
filtro_trilha_pop = ['Pop', 'Trilha Sonora']

df_dados_music_trilha = df_playlist[df_playlist['genero'].isin(filtro_trilha_pop)]

display(df_dados_music_trilha)
```
