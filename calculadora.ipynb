```python
# CRIADO POR LEANDRO SOZINHO#
while True:
  numb1 = int(input('\nDigite o primeiro valor: '))
  numb2 = int(input('Digite o segundo valor: '))

  operator = int(input('\n1: Soma \n'
   '2: Subtração \n'
   '3: Divisão \n'
   '4: Multiplicação \n'
   '5: Potenciação \n'
   '0: Sair \nNumero do operador:'))

  if(operator == 0):
    print('Saindo...')
    break

  match operator:
    case 1:
      print(f'\nA soma de {numb1} + {numb2} é igual a: {numb1 + numb2}')
    case 2:
      print(f'\nA Subtração de {numb1} - {numb2} é igual a: {numb1 - numb2}')
    case 3:
      if operator == 0:
        print('NÃO SE PODE DIVIDIR POR ZERO!!')
      else:
        print(f'\nA Divisão de {numb1} / {numb2} é igual a: {numb1 / numb2}')
    case 4:
      print(f'\nA Multiplicação de {numb1} * {numb2} é igual a: {numb1 * numb2}')
    case 5:
      print(f'\nA Potenciação de {numb1} ** {numb2} é igual a: {numb1 ** numb2}')
    case _:
      print('VALOR INVÁLIDO ')

  continue_program = input('\nDeseja continuar? Sim | Não: ').lower().strip()

  if(continue_program == 'sim'):
    continue
  else:
    print('\nPrograma finalizado com sucesso...')
    break
```

```python
#FAZENDO COM PROFESSOR APRENDENDO FUNCTIONS#

#iniciando uma função
def calculator(operand1, operand2, operator):

  if operator == '+':
    return operand1 + operand2
  elif operator == '-':
    return operand1 - operand2
  elif operator == '*':
    return operand1 * operand2
  elif operator == '/':
    if operator == 0:
     return 'Error, não pode dividir por (0)'
    else:
      return operand1 / operand2
  else:
    return 'Escolha um operador válido'
```

```python
#Simulator calculator
operand1 = float(input('Informe um número: '))
operand2 = float(input('Informe outro número: '))
operator = input('/nInforme o operador: ')

calculator(operand1, operand2, operator)
```
