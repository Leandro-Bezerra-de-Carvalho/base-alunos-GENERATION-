```python
#Exercicio 03 - Faça um programa que peça dois números e imprima a soma:
n1 = int(input('Digite um numero: '))
n2 = int(input('Digite outro numero: '))

print(f'A SOMA É: {n1 + n2}')
```

```python
#Exercicio 04 - Faça um programa que peça as 4 notas bimestrais e mostre a média.
num1 = int(input('Digite 1 nota: '))
num2 = int(input('Digite 2 nota: '))
num3 = int(input('Digite 3 nota: '))
num4 = int(input('Digite 4 nota: '))

media = (num1 + num2 + num3 + num4) / 4

print(f'A média das notas é: {media}')
```
