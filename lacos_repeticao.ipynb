## Laços de repetição

- **for** - Iterar sequências
- **while** - Repetir enquanto condição
- **enumerate** - Índice + valor
- **break/continue** - Controlar fluxo

```python
balance = 0

while True:
  values_products = float(input('Informe o valor do item: '))

  suma = balance + values_products

  ask = int(input('Tem um novo valor a informar? (1 - Sim |2 - Não)'))

  if(ask == 1):
    print('Digite novo valor')
  else:
    print(f'O seu saldo esta no valor de: {suma}')
    break
```

```python
# Cadastrar um produto consumido
products = []

while True:
  new_products = input('Informe o novo produto: ')
  products.append(new_products)

  ask = int(input('Você tem um novo produto? [1: Sim | 2: Não] '))

  if(ask == 1):
    print('Legal! Vamos adicionar \n')
  else:

    for i in products:
      print(f'\n A sua compra foi: {i}')

    break
```

```python
# Cadastrar o produto e o preço
# Somar o valor total da compra

list_products = []
list_price = []

while True:
  new_productsList = input('Digite o nome do produto: ')
  price_products = float(input('Qual é o valor desse produto? R$'))

  list_products.append(new_productsList)
  list_price.append(price_products)

  askNew = int(input('Deseja cadastrar um novo produto? [1: Sim | 2: Não])'))

  if askNew == 1:
    print('\n ----------- Vamos cadastrar um novo produto -----------')

  else:
    #O zip ajunta duas coisas diferentes para ele trabalhar
    for i, j in zip(list_products, list_price):
      print(f'Você comprou {i} de R${j}')

    total_price = sum(list_price)
    print(f'\n A sua compra no total deu: R${total_price:.2f}')
    break
```

```python

#O programa deve realizar uma contagem de 1 até um número limite definido pelo usuário. Para
#cada número na contagem, as seguintes regras de divisibilidade devem ser aplicadas:
#● Se o número for múltiplo de 3, o programa deve imprimir "Fizz".
#● Se o número for múltiplo de 5, o programa deve imprimir "Buzz".
#● Se o número for múltiplo de ambos (3 e 5 simultaneamente), o programa deve imprimir
#"FizzBuzz".
#● Caso contrário, o programa deve imprimir o próprio número.

while True:
  limite = input('Digite o numero limite ou escreva SAIR: ')

  #Verificação de saida
  if limite.lower() == 'sair':
    print('Saindo do jogo...')
    print('Jogo encerrado! ')
    break

  #Verificação se é numero e transformar
  if not limite.isdigit():
    print("Digite um número válido!")
    # Volta para o início do while sem executar o restante do código.
    continue

  # Converte a string para inteiro
  limite = int(limite)

  for i in range(1, limite):
    if i % 3 == 0 and i % 5 == 0:
      print('FizzBuzz')
    elif i % 3 == 0:
      print('Fizz')
    elif i % 5 == 0:
      print('Buzz')
    else:
      print(i)
```
