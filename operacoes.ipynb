```python
# Desafio - Remove velues the list
list_pets = ['Meg','Duda','Pandora','Joy','Garoa','Mini','Soneca','Honda','Lucky','Bingo']

print(f'Lista antiga: {list_pets}')

# Sortear uma posição
import random

sorteado = random.choice(list_pets)
print(f'Valor sorteado: {sorteado}')

# Descobre a posição do pet sorteado
indice = list_pets.index(sorteado)

# Remove usando pop()
removido = list_pets.pop(indice)
print(f'Pet removido: {removido}')

print(f'Lista atual: {list_pets}')
```

```python
# OPERADORES MATEMATICOS

# +  Soma
# -  Subtração
# *  Multiplicação
# /  Divisão
# // Divisão inteira
# %  Resto da divisão
# ** Potência

num1 = input('Digite um numero: ')
num2 = input('Digite outro numero: ')

if num1 > num2:
  print(f'O numero {num1} é maior que o numero {num2}')
elif num1 == num2:
  print(f'Os números são iguais')
else:
  print(f'O numero {num2} é maior do que o número {num1}')
```

```python
# Se o resto da divisão por 2 for 0, o número é par.
# Caso contrário, o número é ímpar.

operando1 = int(input('Digite um valor: '))

operando2 = int(input('\n2 Digite outro valor: '))

resultado = operando1 + operando2

if resultado % 2 == 0:
    print('\n2 Par')
else:
    print('\n2 Ímpar')

print(resultado % 2)
```
