```python
import pandas as pd

vendas = '/content/dados/vendas.csv'
clientes = '/content/dados/clientes.csv'
dados = '/content/dados/dados.csv'
produto = '/content/dados/produtos.csv'

df_vendas= pd.read_csv(vendas)
df_clientes = pd.read_csv(clientes)

# MERGE THE VENDAS AND CLIENTES
df_vendas_clientes = pd.merge(df_vendas, df_clientes,
                              left_on='Cliente',
                              right_on= 'Id Cliente',
                              how = 'inner')

display(df_vendas_clientes)
```

```python
#DESAFIO -> QUANTO O CLIENTE GASTOU EM COMPRA

# Descobrir onde estar o valor do produto
df_produtos = pd.read_csv(produto)

# mesclar tabela df_vendas_clientes com a tabela que tem o valor
df_vendas_produtos = pd.merge(df_vendas_clientes, df_produtos,
                              left_on = 'Produto',
                              right_on= 'Produto',
                              how = 'inner')

# criar novo campo, para mostrar o total gasto na compra

df_vendas_produtos['Total Gasto'] = (df_vendas_produtos['Valor'] * df_vendas_produtos['Quantidade'])

# Display
display(df_vendas_produtos)
```
